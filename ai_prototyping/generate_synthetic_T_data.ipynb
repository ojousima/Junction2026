```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

dt = 5  # seconds
t = np.arange(0, 90 * 60, dt)
minute = t / 60

ambient = 2 + 0.2 * np.sin(t / 900)

# Approximate steady heating-mode coil temperature
coil = ambient - 7.0

# Two synthetic defrost events
events = [(30, 5), (65, 11)]  # start minute, duration

state = np.array(["heating"] * len(t), dtype=object)

for start, duration in events:
    active = (minute >= start) & (minute < start + duration)
    phase = (minute[active] - start) / duration
    coil[active] = -5 + 16 * np.sin(np.pi * phase)
    state[active] = "defrost"

rng = np.random.default_rng(42)
coil += rng.normal(0, 0.15, len(coil))

df = pd.DataFrame({
    "time_s": t,
    "ambient_c": ambient,
    "coil_c": coil,
    "state": state
})
df.to_csv("heatpump_demo.csv", index=False)

plt.plot(minute, ambient, label="Ambient")
plt.plot(minute, coil, label="Coil")
plt.xlabel("Time (minutes)")
plt.ylabel("Temperature (°C)")
plt.legend()
plt.show()
```
